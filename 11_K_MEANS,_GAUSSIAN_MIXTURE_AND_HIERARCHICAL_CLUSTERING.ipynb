```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, adjusted_rand_score
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import load_iris

# ---- Load dataset ----
iris = load_iris()
X = iris.data
y_true = iris.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# =========================================================
# 1. K-Means Clustering
# =========================================================
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

print("=" * 60)
print("K-MEANS CLUSTERING")
print("=" * 60)
print(f"Cluster centers (scaled space):\n{kmeans.cluster_centers_}")
print(f"Silhouette Score: {silhouette_score(X_scaled, kmeans_labels):.4f}")
print(f"Adjusted Rand Index vs true labels: {adjusted_rand_score(y_true, kmeans_labels):.4f}")
```

```python
# =========================================================
# 2. Gaussian Mixture Model (mixtures of Gaussians)
# =========================================================
gmm = GaussianMixture(n_components=3, random_state=42)
gmm_labels = gmm.fit_predict(X_scaled)

print("\n" + "=" * 60)
print("GAUSSIAN MIXTURE MODEL (GMM)")
print("=" * 60)
print(f"Converged: {gmm.converged_}")
print(f"Silhouette Score: {silhouette_score(X_scaled, gmm_labels):.4f}")
print(f"Adjusted Rand Index vs true labels: {adjusted_rand_score(y_true, gmm_labels):.4f}")
```

```python
# =========================================================
# 3. Hierarchical (Agglomerative) Clustering
# =========================================================
hier = AgglomerativeClustering(n_clusters=3, linkage='ward')
hier_labels = hier.fit_predict(X_scaled)

print("\n" + "=" * 60)
print("HIERARCHICAL (AGGLOMERATIVE) CLUSTERING")
print("=" * 60)
print(f"Silhouette Score: {silhouette_score(X_scaled, hier_labels):.4f}")
print(f"Adjusted Rand Index vs true labels: {adjusted_rand_score(y_true, hier_labels):.4f}")
```

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use("Agg")

# ---- Dendrogram ----
plt.figure(figsize=(10, 6))
Z = linkage(X_scaled, method='ward')
dendrogram(Z, truncate_mode='level', p=5)
plt.title("Hierarchical Clustering Dendrogram (Ward linkage)")
plt.xlabel("Sample index / cluster size")
plt.ylabel("Distance")
plt.tight_layout()
plt.savefig("exp5_dendrogram.png", dpi=150)
print("\nDendrogram saved as exp5_dendrogram.png")

# ---- Scatter plot comparison (first two features) ----
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
for ax, labels, title in zip(axes, [kmeans_labels, gmm_labels, hier_labels],
                             ["K-Means", "Gaussian Mixture", "Hierarchical"]):
    ax.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='viridis', s=30)
    ax.set_title(title)
    ax.set_xlabel("Sepal length (scaled)")
    ax.set_ylabel("Sepal width (scaled)")
plt.tight_layout()
plt.savefig("exp5_clustering_comparison.png", dpi=150)
print("Comparison plot saved as exp5_clustering_comparison.png")
```
