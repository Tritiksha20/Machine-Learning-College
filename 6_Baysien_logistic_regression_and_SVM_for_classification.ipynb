```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix
)

# ---- Load dataset ----
df = pd.read_csv("/content/breast_cancer.csv")
X = df.drop(columns=['target', 'diagnosis'])
y = df['target'] # 0 = malignant, 1 = benign

# ---- Train-test split ----
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# ---- Feature scaling ----
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
bayes_log_reg = LogisticRegression(
    penalty='l2', C=1.0, solver='lbfgs', max_iter=5000
)
bayes_log_reg.fit(X_train_scaled, y_train)
y_pred_blr = bayes_log_reg.predict(X_test_scaled)
y_proba_blr = bayes_log_reg.predict_proba(X_test_scaled)

print("=" * 60)

print("BAYESIAN LOGISTIC REGRESSION (MAP estimate)")
print("=" * 60)
print(f"Accuracy: {accuracy_score(y_test, y_pred_blr):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_blr, target_names=['malignant', 'benign']))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_blr))
print("\nSample posterior predictive probabilities (first 5 test points):")
print(np.round(y_proba_blr[:5], 3))
```

```python
svm_clf = SVC(kernel='rbf', C=1.0, gamma='scale', probability=True, random_state=42)
svm_clf.fit(X_train_scaled, y_train)
y_pred_svm = svm_clf.predict(X_test_scaled)

print("\n" + "=" * 60)
print("SUPPORT VECTOR MACHINE (RBF kernel)")
print("=" * 60)
print(f"Accuracy: {accuracy_score(y_test, y_pred_svm):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_svm, target_names=['malignant', 'benign']))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_svm))
```

```python
print("\n" + "=" * 60)

print("MODEL COMPARISON")
print("=" * 60)
print(f"Bayesian Logistic Regression Accuracy: {accuracy_score(y_test, y_pred_blr):.4f}")
print(f"SVM Accuracy : {accuracy_score(y_test, y_pred_svm):.4f}")
```
