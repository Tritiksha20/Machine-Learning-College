```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 1. Generate Synthetic Dataset
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.70, random_state=42)

# 2. Scale Features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply K-Means Clustering (K=3)
kmeans = KMeans(n_clusters=3, init='k-means++', n_init=10, max_iter=300, random_state=42)
cluster_labels = kmeans.fit_predict(X_scaled)

# 4. Evaluate Performance
score = silhouette_score(X_scaled, cluster_labels)
centroids = kmeans.cluster_centers_

# 5. Output Results
print("=" * 50)
print("EXPERIMENT 2: K-MEANS CLUSTERING ALGORITHM")
print("=" * 50)
print(f"Number of Clusters (K): {kmeans.n_clusters}")
print(f"Silhouette Score: {score:.4f}\n")
print("Cluster Centroids (Standardized Scale):")
for i, center in enumerate(centroids):
    print(f"Cluster {i} Center: Feature1 = {center[0]:.3f}, Feature2 = {center[1]:.3f}")
# Display first 10 cluster assignments
df_results = pd.DataFrame(X_scaled, columns=['Feature_1', 'Feature_2'])
df_results['Assigned_Cluster'] = cluster_labels
print("\nSample Data Points with Cluster Assignments:")
print(df_results.head(10))
```
