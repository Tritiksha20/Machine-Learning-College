```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris

# ---- Load dataset ----
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target_names[iris.target]
X = df.drop(columns=['species'])
y = df['species']

# ---- Standardize the features ----
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# =========================================================
# PCA with all components (to inspect explained variance)
# =========================================================
pca_full = PCA()
pca_full.fit(X_scaled)

print("=" * 60)
print("Explained variance ratio per component:")
print("=" * 60)
for i, ratio in enumerate(pca_full.explained_variance_ratio_, start=1):
    print(f"PC{i}: {ratio:.4f}")

print("\nCumulative explained variance:")
print(np.cumsum(pca_full.explained_variance_ratio_))
```

```python
# =========================================================
# Reduce to 2 principal components for visualization
# =========================================================
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("\n" + "=" * 60)
print(f"Total variance explained by 2 components: {sum(pca.explained_variance_ratio_):.4f}")
print("=" * 60)

pca_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
pca_df['species'] = y
print("\nFirst 5 rows of PCA-transformed data:")
print(pca_df.head())
```

```python
# ---- Scree plot ----
plt.figure(figsize=(7, 5))
plt.plot(range(1, len(pca_full.explained_variance_ratio_) + 1),
         pca_full.explained_variance_ratio_, marker='o', linestyle='--')
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("Scree Plot")
plt.tight_layout()
plt.savefig("exp6_scree_plot.png", dpi=150)
```

```python
# ---- 2D scatter plot of PCA components ----
plt.figure(figsize=(8, 6))
for species in pca_df['species'].unique():
    subset = pca_df[pca_df['species'] == species]
    plt.scatter(subset['PC1'], subset['PC2'], label=species, s=40)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Iris Dataset (2 Components)")
plt.legend()
plt.tight_layout()
plt.savefig("exp6_pca_scatter.png", dpi=150)
print("\nPlots saved: exp6_scree_plot.png, exp6_pca_scatter.png")
```
