```python
import pandas as pd
df = pd.read_csv("iris (1).csv")
print("=" * 60)
print("Dataset Info:")
print("=" * 60)
df.info()

print("\n" + "=" * 60)
print("Statistical Summary (numerical columns):")
print("=" * 60)
print(df.describe())

print("\n" + "=" * 60)
print("Statistical Summary (all columns, including categorical):")
print("=" * 60)
print(df.describe(include='all'))

print("\n" + "=" * 60)
print("Missing values per column:")
print("=" * 60)

print(df.isnull().sum())

print("\n" + "=" * 60)
print("Count of each species (class distribution):")
print("=" * 60)
print(df['species'].value_counts())

print("\n" + "=" * 60)
print("Correlation matrix (numerical features):")
print("=" * 60)
numeric_df = df.select_dtypes(include='number')
print(numeric_df.corr())

print("\n" + "=" * 60)
print("Variance of each numerical feature:")
print("=" * 60)
print(numeric_df.var())

print("\n" + "=" * 60)
print("Skewness of each numerical feature:")
print("=" * 60)
print(numeric_df.skew())
```
